```python
import math as m
import pandas as pd
import plotly.express as px
import numpy as np
import msicpe.mana as mana
import scipy.signal as signal
import msicpe.san as san


def tracer(abscisse, courbes, nom_abscisse="t", titre=""):
    donnees = pd.DataFrame({nom_abscisse: abscisse, **courbes})
    fig = px.line(donnees, x=nom_abscisse, y=list(courbes.keys()), title=titre)
    fig.show()

t=np.linspace(-10,10,3000)    # CORRECTION : ne plus utiliser ce t, chaque signal a son t (t_a, t_e...)
nu= np.linspace(-10, 10,3000)

#nbre d'harmonique 
H= 10    # CORRECTION : 5 -> 10 (H = 5 ne voit que le bas du spectre)
a_6=np.load("a_6.npz")
s_a= a_6['s']
t_a = a_6['t']

e_6=np.load("e_6.npz")
s_e = e_6['s']
t_e = e_6['t']

o_6=np.load("o_6.npz")
s_o = o_6['s']
t_o = o_6['t']

u_6=np.load("u_6.npz")
s_u = u_6['s']
t_u = u_6['t']


# CORRECTION : ces listes sont recalculees plus bas, apres detect_fondamentale,
# car detect_pics veut les frequences des harmoniques en Hz : f0, 2f0, ..., (H+1)f0
H_a = [i for i in range(1,H+1)]
H_e = [i for i in range(1,H+1)]
H_o = [i for i in range(1,H+1)]
H_u = [i for i in range(1,H+1)]
```

```python
nu, A = san.trans_fourier(s_a - np.mean(s_a),t_a)   # CORRECTION : t_a au lieu de t + retrait de la moyenne
A = np.abs(A)/np.max(np.abs(A))                    # CORRECTION : module + normalisation
nu_f =san.detect_fondamentale (A,nu)               # CORRECTION : A au lieu de s_a
H_a = [k*nu_f for k in range(1,H+2)]               # CORRECTION : harmoniques en Hz
freq_a, nu_a= san.detect_pics(A,H_a,nu)            # CORRECTION : A au lieu de s_a
nu_a = np.log10(nu_a)                              # CORRECTION : le log10 n'etait pas pris
print("Valeurs en log10 de l'amplitude ",nu_a, "a pour fréquence :",freq_a)   # CORRECTION : npfreq_a -> freq_a

nu, E = san.trans_fourier(s_e - np.mean(s_e),t_e)
E = np.abs(E)/np.max(np.abs(E))
nu_f =san.detect_fondamentale (E,nu)
H_e = [k*nu_f for k in range(1,H+2)]
freq_e, nu_e = san.detect_pics(E,H_e,nu)
nu_e = np.log10(nu_e)
print("Valeurs en log10 de l'amplitude ",nu_e, "e pour fréquence :", freq_e)

nu, O = san.trans_fourier(s_o - np.mean(s_o),t_o)
O = np.abs(O)/np.max(np.abs(O))
nu_f =san.detect_fondamentale (O,nu)
H_o = [k*nu_f for k in range(1,H+2)]
freq_o, nu_o = san.detect_pics(O,H_o,nu)
nu_o = np.log10(nu_o)
print("Valeurs en log10 de l'amplitude ",nu_o, "o pour fréquence :", freq_o)

nu, U = san.trans_fourier(s_u - np.mean(s_u),t_u)
U = np.abs(U)/np.max(np.abs(U))
nu_f =san.detect_fondamentale (U,nu)
H_u = [k*nu_f for k in range(1,H+2)]
freq_u, nu_u = san.detect_pics(U,H_u,nu)
nu_u = np.log10(nu_u)
print("Valeurs en log10 de l'amplitude ",nu_u, "u pour fréquence :", freq_u)

```
