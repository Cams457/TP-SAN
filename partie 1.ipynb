```python
import math as m
import pandas as pd
import plotly.express as px
import numpy as np
import msicpe.mana as mana
import scipy.signal as signal
import msicpe.san as san


def tracer(abscisse, courbes, nom_abscisse="t", titre=""):
    donnees = pd.DataFrame({nom_abscisse: abscisse, **courbes})
    fig = px.line(donnees, x=nom_abscisse, y=list(courbes.keys()), title=titre)
    fig.show()

t=np.linspace(-10,10,3000)
nu= np.linspace(-10, 10,3000)

```

```python
violin_dict = np.load('violin_C.npz')
s = violin_dict['s']
t = violin_dict['t']

nu, S=san.trans_fourier(s,t)

tracer(nu, {"S": np.imag(S)}, nom_abscisse="nu", titre="Spectre de la note C")
tracer(t, {"s": s}, nom_abscisse="t", titre="Signal de la note C")
```
